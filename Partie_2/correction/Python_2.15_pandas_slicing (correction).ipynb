# Mission 5

Rename the columns [of this dataset](https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/ufo.csv), try using the following 2 different methods to achieve this:
- Setting column names directly when reading the file
- Editing after creating the dataframe ([This resource could be useful](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.rename.html))

```
mission5_data = pd.read_csv("https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/ufo.csv",sep = ",")
mission5_data
```

```
# Setting column name directly when reading the file
mission5_data = pd.read_csv("https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/ufo.csv",
                            sep = ",",
                            names = ['Ville','Couleur','Forme','Etat','Temps'],
                            skiprows = 1)
mission5_data
```

```
# Editing after creating the dataframe
mission5_data = pd.read_csv("https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/ufo.csv",sep = ",")
mission5_data = mission5_data.rename(columns = {'City':'Ville','Colors Reported':'Couleur','Shape Reported':'Forme','State':'Etat','Time':'Temps'})
mission5_data
```
