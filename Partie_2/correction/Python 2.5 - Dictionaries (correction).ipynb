# Mission 1

Below is a dictionary containing the number of French campuses that offer each kind of Wild Code School course.

wildcodeschool = {
   "data_analyst": 4,
   "data_scientist": 2,
   "dev_web": "13"
}

Find a way to only display the keys that have int values.

```python
wildcodeschool =	{
  "data_analyst": 4,
  "data_scientist": 2,
  "dev_web": "13"
}
```

```python
# First we have to iterate over the dictionary to get the key and value pairs
# Then for each these pairs check if the value is of type 'int' (integer)
```

```python
## Method 1

for key, value in wildcodeschool.items():
    if type(value) == int :                 # check if value's type equals int
        print(key)
```

```python
## Method 2

for key, value in wildcodeschool.items():
    if type(value) in [int] :              # check value's type is in a list of desired types
        print(key)
```

# Mission 2

For the same dictionary, find a way to display all the values that have a key that contains the letter "e".

```python
## Method 1

# By default, if we use a single iterator (we use k here, but we could use i, j, dad, mum...) only the keys are extracted from the dictionary

for key in wildcodeschool:
    if 'e' in key :                  # check if the 'e' letter is in the current dictionary key
        print(wildcodeschool[key])   # get and print the value corresponding to the current key
```

```python
## Method 2

# We will do as we did for Mission 1 :
# Loop over the dictionary to get the key-value pairs - check if the letter 'e' is found in the key - print the value if 'e' is present in the key

## Method 1

for key, value in wildcodeschool.items():
    if 'e' in key:                 # check if the 'e' letter is in the current dictionary key
        print(value)               # print the current value (corresponding to the current key)
```

# Mission 3

Within the same dictionary, change the type of the value of the "dev_web" key to be an int. Then add a new key-value pair which will be "total_campus" : 19, using a **dictionary method**.

**Change the value of "dev_web" to be and int :**

```python
# We will use the int() function for that

wildcodeschool['dev_web'] = int(wildcodeschool['dev_web'])   # int() will convert the str type to int type and the value will be assigned to the 'dev_web' key
print(wildcodeschool)                                        # print the dictionary to check if things worked fine
```

**Add a new key-value pair which will be "total_campus" : 19, using a dictionary method**

```python
## Method 1
# We will add a new key-value pair to the dictionary using the update() method which is a dictionary method

wildcodeschool.update({"total_campus": 19})        # add a new key-value pair
print(wildcodeschool)                              # print the dictionary to check if things worked fine
```

```python
## Method 2
# We can also create a new key-value pair using the following method

wildcodeschool['total_campus'] = 19     # Assign the value 19 to the key 'total_campus'. Since both elements doesn't exist yet, they will be added
print(wildcodeschool)                   # print the dictionary to check if things worked fine
```

# Mission 6

Create a script that returns the same elements as the **values()** method, as a list.

Example: for `{'a': 1, 'b': 2, 'c': 3}` it will create `[1, 2, 3]`

```python
def values(a_dict):
    values_list = []
    for key in a_dict:
        values_list.append(a_dict[key])
    return values_list
```

```python
# Check if our script works fine
dico = {'a': 1, 'b': 2, 'c': 3}
print(values(dico))
```

```python
# Check if our script works fine
print(values(wildcodeschool))
```
