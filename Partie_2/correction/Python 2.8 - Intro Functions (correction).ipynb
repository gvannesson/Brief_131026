# Mission 2

Create a function that returns the sum of two numbers.  
The user of the function must be able to sum two numbers of his choice, when the function is executed.  
You must not use `input`.  


```python
def new_sum(a_number, another_number):
  return a_number + another_number

new_sum(2,5)
```

Question:

 When defining a function, a parameter is the name given to the elements contained between the brackets.

# Mission 4

Create a function that returns 2 objects when executed:
* An object that will be the product of two numbers chosen by the user of the function.
* A word chosen by the user of the function.

```python
def sum_and_word(a,b, word):
  product = a * b
  return product, word

sum_and_word(6,7, "gondille")
```

Your result must be two objects that are in brackets.  
This is a tuple.

Question:  
Why is the returned object a tuple that contains two objects?  

Answer:  
The function cannot return two or more object at once. So, the result can be pack in one object, which is a tuple.

# Mission 5

Create a function that returns the sum of two numbers when executed.  
Also, your function must return a sentence that simply indicates if the sum of the two numbers chosen by the user is:
* less than 10
* between 10 and 50
* greater than 50  
Again, you must use `return`.

```python
def new_sum_3(a,b):
  sum_of = a + b
  if sum_of < 10:
    return "The sum of the two numbers is less than 10."
  elif sum_of >= 10 and sum_of <= 50:
    return "The sum of the two numbers is between 10 and 50."
  else:
    return "The sum of the two numbers is greater than 50."
```

```python
new_sum_3(2, 5)
```

```python
new_sum_3(12,15)
```

```python
new_sum_3(3, 60)
```
