# Mission 2

Create a "fruit_vendors" dataframe that reflects the below example **using a dictionary** data structure

![](https://i.imgur.com/CHPn7ZF.png)

[This resource could help you.](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.from_dict.html)

```
fruit_vendors = pd.DataFrame({
    'Apples': {'2017 Sales': 35, '2018 Sales': 41},
    'Bananas': {'2017 Sales': 21, '2018 Sales': 34},
})
fruit_vendors
```

# Mission 3

Use a loop to display the values from each column in the dataframe from mission 2.


```
# There are many, many ways to do this, here's one
for column in fruit_vendors.columns:
  print(column, fruit_vendors[column].values)
```

# Mission 4

Delete the first row of the dataframe from mission 2

[This resource could help you.](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.drop.html)

```
# By index, if we know the specific index to drop
fruit_vendors.drop('2017 Sales')

# By location ("drop the first row")
fruit_vendors.drop(fruit_vendors.index[0])
```

```
# Note: the above didn't edit the table!
# You'd have to use inplace=True for the changes to persist, or reassign the table!
fruit_vendors
```

# Mission 5

Delete the first column from the dataframe from mission 2.


```
# By column name, if we know the specific one to drop
fruit_vendors.drop(columns='Apples')

# By location ("drop the first colums")
fruit_vendors.drop(columns=fruit_vendors.columns[0])
```
