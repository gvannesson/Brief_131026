```python
import math

# Variables
PI = math.pi
E = math.e

# Fonctions de base
def carre(x):
    return x ** 2

def cube(x):
    return x ** 3

def factorielle(n):
    if n == 0 or n == 1:
        return 1
    return n * factorielle(n - 1)

def valeur_absolue(x):
    return abs(x)

def maximum(liste):
    return max(liste)

def minimum(liste):
    return min(liste)

def moyenne(liste):
    return sum(liste) / len(liste)

def ecart_type(liste):
    moy = moyenne(liste)
    variance = sum((x - moy) ** 2 for x in liste) / len(liste)
    return math.sqrt(variance)

def mediane(liste):
    liste_triee = sorted(liste)
    n = len(liste_triee)
    if n % 2 == 0:
        return (liste_triee[n//2 - 1] + liste_triee[n//2]) / 2
    else:
        return liste_triee[n//2]

# Nouvelles fonctions
def somme_cumulee(liste):
    """Calcule la somme cumulée d'une liste."""
    return [sum(liste[:i+1]) for i in range(len(liste))]

def compter_occurrences(liste):
    """Retourne un dictionnaire avec le nombre d'occurrences de chaque élément dans la liste."""
    return {x: liste.count(x) for x in set(liste)}

def inverser_dict(dictionnaire):
    """Inverse les clés et les valeurs d'un dictionnaire."""
    return {v: k for k, v in dictionnaire.items()}

def filtrer_pairs(liste):
    """Retourne une nouvelle liste contenant uniquement les nombres pairs de la liste d'entrée."""
    return [x for x in liste if x % 2 == 0]

def celsius_to_fahrenheit(celsius):
    """Convertit une température de Celsius en Fahrenheit."""
    return (celsius * 9/5) + 32

def trouver_doublons(liste):
    """Retourne une liste des éléments qui apparaissent plus d'une fois dans la liste d'entrée."""
    return [x for x in set(liste) if liste.count(x) > 1]

def calculer_moyenne_ponderee(valeurs, poids):
    """Calcule la moyenne pondérée d'une liste de valeurs avec une liste de poids correspondante."""
    return sum(v * p for v, p in zip(valeurs, poids)) / sum(poids)

def analyser_texte(texte):
    """Analyse un texte et retourne diverses statistiques."""
    mots = texte.split()
    return {
        "nombre_mots": len(mots),
        "nombre_caracteres_avec_espaces": len(texte),
        "nombre_caracteres_sans_espaces": len(texte.replace(" ", "")),
        "mot_plus_long": max(mots, key=len),
        "longueur_moyenne_mots": sum(len(mot) for mot in mots) / len(mots)
    }
```
