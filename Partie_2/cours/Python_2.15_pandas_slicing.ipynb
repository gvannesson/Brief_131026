#### Comment importer un fichier ?
Pandas permet d'importer facilement un fichier. Par exemple, pour un fichier CSV (Comma Separated Values), nous utilisons la méthode `read_csv` en indiquant que le séparateur est une virgule.

```python
import pandas as pd
# You can open a file with different "read" method with pandas

pd.read_csv("https://forge.scilab.org/index.php/p/rdataset/source/file/master/csv/datasets/iris.csv", 
            sep = ",")
# There are lots of arguments, here we precise the separator symbol ","
```

- `sep` : pour le séparateur
- `header` et `names` : pour le nom des en-têtes de colonne
- `index_col` : pour le nom des en-têtes de ligne
- `skiprows` : pour sauter des lignes s'il y a des commentaires en début de fichier

#### Sélectionner une ligne ou une colonne : slicing

Pour sélectionner une partie (= un slice) d'un DataFrame, il y a 2 méthodes principales :

**Méthode iloc**
Tu sélectionnes les lignes et les colonnes en comptant à partir de 0, comme tu as l'habitude de le faire en python, et en NumPy.

```python
df.iloc[0 , [-2:]] # Ici on sélectionne la première ligne (0) et les 2 dernières colonnes
```

**Méthode loc**
Tu sélectionnes les lignes et les colonnes par leurs en-tête, c'est assez proche d'un tableur.

```python
df.loc[:, "Amount"] # Ici on affiche toutes les lignes de la colonne "Amount"
```
