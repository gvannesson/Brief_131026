#### Filtrer des données

Avec Pandas, tu peux filtrer les données que tu souhaites afficher suivant leurs valeurs.

Reprenons notre exemple de fruits :

|  | apple | pear |
| --- | ----- | ---- |
| **last year** | 1 | 2 |
| **last month** | 3 | 4 |
| **today** | 5 | 6 |


**Tests logiques**
Je peux appliquer un test logique à une Série (= une colonne) ou bien à tout le DataFrame.

Test logique pour le DataFrame entier, j'obtiens un DataFrame rempli de booléens :

```python
df == 3
```

|  | apple | pear |
| --- | ----- | ---- |
| **last year** | False | False |
| **last month** | True | False |
| **today** | False | False |



Test logique pour une Série, j'obtiens une Série remplie de booléens :

```python
df['apple'] >= 3
```

|  | apple |
| --- | ----- |
| **last year** | False |
| **last month** | True |
| **today** | True |



**Filtrer avec un .loc**
Rappelons-nous que la syntaxe du .loc permet de filtrer sur des lignes en fonction de leur en-tête de ligne. Comme ici :

```python
df.loc[['last month', 'today']]
```

|  | apple | pear |
| --- | ----- | ---- |
| last month | 3 | 4 |
| today | 5 | 6 |



Je vais maintenant faire le même filtre, mais en indiquant au .loc de ne conserver que les lignes pour lesquelles ma Série renvoyait des "True".

```python
df.loc[ df['apple'] >= 3 ]
```

|  | apple | pear |
| --- | ----- | ---- |
| last month | 3 | 4 |
| today | 5 | 6 |

```story
**Attention**
Pour mettre plusieurs critères, Pandas n'utilise pas la syntaxe standard Python. Il faut utiliser la syntaxe suivante :

* `or` doit être remplacé par `|`
* `and` doit être remplacé par `&`

Par exemple :
`df.loc[ (df['apple'] >= 3) &  (df['pear'] < 6)]`

### Ressources


#### Un toujours très bon "cheatsheet"
https://pandas.pydata.org/Pandas_Cheat_Sheet.pdf
