#### Des datas et des tableaux

Pandas sert à manipuler des tableaux que nous appellerons `DataFrame`.
Un **dataframe** est un **tableau de données**. Les données sont rangées par colonne et ligne, exactement comme dans une feuille Excel ou une base de données SQL. Par exemple en Python, une colonne d'un dataframe, appelée **Series**, peut être considérée comme une liste, ce qui donne accès à tous les mécanismes que vous connaissez sur les listes.

![](https://storage.googleapis.com/lds-media/images/series-and-dataframe.width-1200.png)


**La configuration sous forme de dataframe est très pratique pour l'analyse de données**. Elle a donc été adoptée par les Data Analysts.

Pandas permet d'effectuer de l'algèbre linéaire, notamment des jointures (comme dans SQL). Nous y reviendrons dans d'autres quêtes plus avancées.

#### Et Pandas dans tout ça ?

![](https://cdn-images-1.medium.com/max/1600/1*6p6nF4_5XpHgcrYRrLYVAw.png)


Techniquement, **Pandas s'appuie sur la bibliothèque NumPy**, que tu connais déjà. Pandas limite les dimensions à 2 (lignes x colonnes), et ajoute un grand nombre d'attributs (en-têtes de colonnes et de lignes, etc...) et de méthodes, qui seront très adaptés au traitement de données.

#### Création de DataFrames

Nous allons commencer dans cette quête par créer des DataFrames manuellement. Plus tard, nous verrons comment importer des données depuis des fichiers ou des bases de données, mais chaque chose en son temps.

```python
import pandas as pd
pd.DataFrame([1,2,42])
```

![ ](https://storage.googleapis.com/quest_editor_uploads/guYDzUISliImEFU8mofQplLwHsdeoh7v.png)

Ici je crée un DataFrame à partir d'une liste. Tu peux créer un DataFrame à partir de toutes les structures de données que tu connais déjà (listes, dictionnaires, arrays NumPy).

Par défaut, les en-têtes de ligne et les en-têtes de colonnes sont numérotés à partir de 0. Pour changer les noms de colonnes, par des libellés plus explicites, tu peux utiliser le paramètre `columns` :

```python
pd.DataFrame([1,2,42], columns=["number"])
```
