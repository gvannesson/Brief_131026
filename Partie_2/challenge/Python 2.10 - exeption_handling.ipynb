You must correct each of the scripts below **using exception handling** and **without modifying lists**.

Script number 1

```python
# Erreur d'origine : ZeroDivisionError quand i vaut 0
list_integers = [1, 12, -45, 6, 27, 0, 24]
for i in list_integers:
  try:
    print(1/i)
  except ZeroDivisionError:
    print("Impossible de diviser par 0")
```

Script number 2

```python
# Erreur d'origine : IndexError, car range(10) dépasse la longueur de la liste (6 éléments)
list_integers = [1, 12, -45, 6, 27, 24]
for i in range(10):
  try:
    print(1/list_integers[i])
  except IndexError:
    print("Fin de la liste atteinte")
    break
```

Script number 3

```python
# Deux erreurs possibles : ZeroDivisionError (le 0) et IndexError (range(10) dépasse la liste)
list_integers = [1, 12, -45, 6, 27, 0, 24]
for i in range(10):
  try:
    print(1/list_integers[i])
  except ZeroDivisionError:
    print("Impossible de diviser par 0")
  except IndexError:
    print("Fin de la liste atteinte")
    break
```

Script number 4

```python
# Erreur d'origine : AttributeError, car 34 est un int, qui n'a pas de méthode .upper()
list_words = ['hello', 'good morning', 'GOOD Evening', 34, 'Bye' ]

for word in list_words:
  try:
    print(word.upper())
  except AttributeError:
    print(f"{word} n'est pas un texte, impossible de le mettre en majuscules")
```
