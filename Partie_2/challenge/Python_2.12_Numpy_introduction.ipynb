```python
# Mission 4 : What does the np.empty() method do? Don't just copy the documentation, explain in your own words here

array_3 = np.empty(4) # The np.empty() method creates an array of a specified shape and type, without initializing the entries. The values in the array will be whatever happens to already exist at that memory location, so they will be random and not necessarily zero.
array_3
"np.empty() creates an array of the given shape without initializing its values. It only reserves the memory, so the array contains whatever data was already there (leftovers from previous operations). The values are not random, just unpredictable. It is slightly faster than np.zeros(), so it is useful when every element will be overwritten right after."
```
