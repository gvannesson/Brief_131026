# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
def demander_nombre(entree_utilisateur):
    while True:
        saisie = input(entree_utilisateur)
        try:
            nombre = float(saisie)
            return nombre
        except ValueError:
            print("Veuillez entrer un nombre valide.")

prix_unitaire = demander_nombre("Entrez le prix unitaire : ")
taux_TVA = demander_nombre("Entrez le taux de TVA (en pourcentage) : ")
nombre_articles = demander_nombre("Entrez le nombre d'articles : ")

montant_total = prix_unitaire * (1 + taux_TVA / 100) * nombre_articles
print(f"Le montant total à payer est : {montant_total:.2f} €")
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
while True:
    duree = input("Entrez la durée en secondes : ")
    try:
        secondes = int(duree)
        if secondes < 0:
            print("Veuillez entrer un nombre positif.")
            continue
        heures = secondes // 3600
        minutes = (secondes % 3600) // 60
        secondes_restantes = secondes % 60
        print(f"{heures} heures, {minutes} minutes et {secondes_restantes} secondes.")
        break
    except ValueError:
        print("Veuillez entrer un nombre entier valide.")
```
