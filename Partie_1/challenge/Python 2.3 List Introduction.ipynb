2. What can the method `.insert()` do ?  
You must answer using the terms 'parameter' and 'argument'.

Answer: 
The .insert() method adds an element to a list at a specific position. It takes two parameters: index (the position where the element is inserted) and object (the element to insert). For example, in fruits.insert(1, "kiwi"), the arguments 1 and "kiwi" are passed to these parameters: "kiwi" is inserted at position 1, and the elements that were from position 1 onwards are shifted one place to the right.
