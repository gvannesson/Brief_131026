# 1. Calcul du montant d'une remise

Un commerçant accorde une remise de 5 % pour tout achat d’un montant compris entre 100 et 500 € et 8 % au-delà.<br>
Ecrivez un programme de calcul du montant de la remise sur un achat donné.<br>
Affichez le montant de la remise.

```python
def demander_nombre(entree_utilisateur):
    while True:
        saisie = input(entree_utilisateur)
        try:
            nombre = float(saisie)
            return nombre
        except ValueError:
            print("Veuillez entrer un nombre valide.")

montant_achat = demander_nombre("Entrez le montant de l'achat : ")

if montant_achat > 100 and montant_achat <= 500:
    remise = 0.05
    print(f"Vous bénéficiez d'une remise de 5% sur votre achat soit, {montant_achat * remise:.2f} euros.")
elif montant_achat > 500:
    remise = 0.08
    print(f"Vous bénéficiez d'une remise de 8% sur votre achat soit, {montant_achat * remise:.2f} euros.")
else:
    print("Vous ne bénéficiez d'aucune remise sur votre achat.")
```

# 2. Savoir si trois entiers sont triés

Ecrivez un programme faisant saisir trois entiers x, y, z à l’utilisateur, et lui indiquer si ces nombres sont dans l’ordre croissant (x <= y <= z).

```python
def demander_nombre_entier(entree_utilisateur):
    while True:
        saisie = input(entree_utilisateur)
        try:
            nombre = int(saisie)
            return nombre
        except ValueError:
            print("Veuillez entrer un nombre entier valide.")   



x = demander_nombre_entier("Entrez un nombre entier : ")
y = demander_nombre_entier("Entrez un nombre entier plus élevé : ")
z = demander_nombre_entier("Entrez un troisième nombre entierplus élevé : ")

if x <= y and y <= z:
    print("Les nombres sont dans l'ordre croissant.")
else:
    print("Les nombres ne sont pas dans l'ordre croissant.")
```
