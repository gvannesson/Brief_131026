```python
import pandas as pd

link = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
df =pd.read_csv(link)
```

```python
texte="[u'Tim Robbins', u'Morgan Freeman', u'Bob Gunton']"
def clean_texte(texte):
    texte= texte.strip("[]")
    texte
    actor_list =texte.split(", ")
    # for actor in actor_list:
    actor_list
    final_actor_list = []
    for actor in actor_list:
        actor = actor.removeprefix("u'")
        actor= actor.removesuffix("'")
        # Les noms contenant une apostrophe sont écrits entre guillemets doubles : u"Peter O'Toole"
        actor = actor.removeprefix('u"')
        actor = actor.removesuffix('"')
        final_actor_list.append(actor)
    return final_actor_list
```

```python
df['actors_list'] = df['actors_list'].apply(clean_texte)
```
