# PIVOT TABLE

## It's like in Excel, but with Pandas

There are 2 syntaxes, choose your preferred :
- As a method : YOUR_DATAFRAME.pivot_table()
- As a function : pd.crosstab()

```python
# Example
df_weather.pivot_table(values = 'MAX_TEMPERATURE_C', index = 'MONTH', aggfunc = 'mean')
```

## Create a Pivot table with :
- Months on rows
- the maximum of Max Temperature on values
- Opinion on columns

```python
# Hop !
df_weather.pivot_table(values='MAX_TEMPERATURE_C', index='MONTH', columns='OPINION', aggfunc='max')
```

## Pivot table with :
- Days on rows
- Precipitation on values
- Months on columns


```python
# Hop !
df_weather.pivot_table(values = 'PRECIP_TOTAL_DAY_MM', index = 'DAY', columns="MONTH")
```

## Pivot table with :
- Months on rows
- the sum of Precipitations on values

**Then display it as a column chart.** (optionnal)

```python
# Hop !
pivot_precip = df_weather.pivot_table(values = 'PRECIP_TOTAL_DAY_MM', index = 'MONTH',  aggfunc='sum')
pivot_precip
```

```python
pivot_precip.plot.bar(
    figsize=(12, 6),
    legend=False,
    rot=0,
    title="Précipitations totales par mois (mm)",
)
```

## Pivot table with :
- Months on rows
- the average of Max Temperature on a first column of values
- the average of Sunhours on a second column of values

```python
# Hop !
sunhour = df_weather.pivot_table(values = ['MAX_TEMPERATURE_C', 'SUNHOUR'], index = 'MONTH',  aggfunc='mean')
sunhour
```

# Chart
Using your favorite library (MatPlotLib, Seaborn, Plotly or what you want), draw a chart from the previous pivot table with :
- Temperature in a blue column chart
- Sunhours in a red line plot

Tip : Once your pivot table created, you can help yourself with the `.reset_index()`  method.

```python
# Your code here
sunhour= sunhour.reset_index()
sunhour
```

```python
import matplotlib.pyplot as plt

sunhour_reset = sunhour.reset_index()

fig, ax1 = plt.subplots(figsize=(12, 6))

# Barres bleues : la température, sur l'axe y de gauche
ax1.bar(sunhour_reset['MONTH'], sunhour_reset['MAX_TEMPERATURE_C'], color='blue')
ax1.set_xlabel('Mois')
ax1.set_ylabel('Température max moyenne (°C)', color='blue')
ax1.set_xticks(sunhour_reset['MONTH'])

# Courbe rouge : l'ensoleillement, sur un second axe y à droite
ax2 = ax1.twinx()
ax2.plot(sunhour_reset['MONTH'], sunhour_reset['SUNHOUR'], color='red', marker='o')
ax2.set_ylabel('Ensoleillement moyen (h)', color='red')

plt.title('Température et ensoleillement moyens par mois (2019)')
plt.show()
```
